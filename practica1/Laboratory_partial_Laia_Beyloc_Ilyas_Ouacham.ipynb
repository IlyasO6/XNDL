# Exercise 1 (1 point)

A scientist studies how a chemical reaction rate depends on temperature. The Arrhenius equation gives the rate constant $k$ as:

$$k = A\, e^{-\frac{E_a}{RT}}$$

where $A$ is the pre-exponential factor, $E_a$ the activation energy, and $R = 8.3145$ J/(mol·K) the universal gas constant.

Experimental data:

$$\{(k, T)\} = \{(0.0021, 300\,\text{K}),\,(0.0063, 320\,\text{K}),\,(0.0192, 340\,\text{K}),\,(0.0580, 360\,\text{K}),\,(0.1650, 380\,\text{K})\}$$

Estimate $A$ and $E_a$ using:

- Direct pseudoinverse (0.5 points)
- Pseudoinverse via SVD (0.5 points)

```python
#  Experimental data
R = 8.3145

k = np.array([0.0021, 0.0063, 0.0192, 0.0580, 0.1650])
T = np.array([300, 320, 340, 360, 380])

y = np.log(k)
X = np.column_stack([np.ones(len(T)), 1.0 / T])
```

```python
# Direct pseudoinverse
pseudoinv = np.linalg.inv(X.T @ X) @ X.T #  formula of the pseudoinverse
theta = pseudoinv @ y

A = np.exp(theta[0])
E_a = -theta[1] * R

labels = ["X", "X_inv", "Theta", "results", "A", "E_a"]
values = [X, pseudoinv, theta, None, A, E_a]
print_results(labels, values)
```

```python
# Pseudoinverse via SVD
X_TX = X.T @ X

eigenvals, eigenvecs = np.linalg.eigh(X_TX)

V = eigenvecs.copy()
delta = np.diag(np.sqrt(eigenvals))
delta_inv = np.linalg.inv(delta)
U = X @ V @ delta_inv

#  check wether the decomposition is correct:
print("="*40, "\n", "Correct decomposition? ", np.allclose(X, U @ delta @ V.T), sep="")

#  now for the pseudoinverse:
pseudoinv2 = V @ delta_inv @ U.T
print("="*40, "\n", "Same pseudoinverse? ", np.allclose(pseudoinv, pseudoinv2), sep="")

theta2 = pseudoinv2 @ y
A2 = np.exp(theta2[0])
E_a2 = -theta2[1] * R

values = [X, pseudoinv2, theta2, None, A2, E_a2]
print_results(labels, values)
```

```python
#  Training data
X_train = np.array([[25, 10.0, 1], [24, 9.5, 1], [12, 5.0, 1], [11, 4.5, 1]]) #  [Ear Size, Tail Length, Bias]
y_train = np.array([1, 1, -1, -1])  # Field=+1, House=-1
#  Validation data
X_val = np.array([[13, 6.0, 1], [27, 11.0, 1]])
y_val = np.array([-1, 1])
```

1. Plot the training and validation data. (0.25 points)

```python
def create_plot(data_label, X, y, axes):
    for label, marker, color in zip([-1, 1], ["v", "^"], ["red", "blue"]):
        species = "Field" if label == 1 else "House"
        axes.scatter(
            X[y == label, 0],
            X[y == label, 1], 
            label=f"Class {species}",
            marker=marker,
            color=color,
        )
    axes.set_xlabel("Ear Size")
    axes.set_ylabel("Tail Length")
    axes.set_title(data_label)
    axes.legend()
    axes.grid(True)

X_train_plot = X_train[:, :2] #  the bias is not plotted, only the variables measured
X_val_plot = X_val[:, :2]

fig, axes = plt.subplots(1, 2, figsize=(10, 5))

create_plot("Training Data", X_train_plot, y_train, axes[0])
create_plot("Validation Data", X_val_plot, y_val, axes[1])

plt.tight_layout()
plt.show()
```

2. Complete the vectorised Perceptron Learning rule below. *Vectorised* means no inner loops over samples. (1 point)

```python
#  made a function to reuse the same setup later
#  added some arguments to the function in case we needed them in the future
def perceptron(X, y, learning_rate = 1, max_iter = 100, verbose = 0):
    w = np.zeros(3)
    mistake = True
    epoch = 0
    while mistake:
        if epoch > max_iter:
            print(f"Couldn't converge in {max_iter} iterations.")
            return w
        # your code here
        if verbose and epoch % 10 == 0:
            print("Epoch", epoch)

        missclassified = y * (X @ w) <= 0
        if any(missclassified): #  we adjust the weights using the wrongly classified data points
            w += learning_rate * np.sum((y[missclassified][:, np.newaxis] * X[missclassified]), axis = 0)
        else:
            mistake = False
        epoch +=1
    print(f"Converged in {epoch} steps.")
    return w, epoch

w, steps = perceptron(X = X_train, y = y_train, verbose=1)
print("w =", w)
```

```python
y_val_pred = np.sign(X_val @ w)

pd.DataFrame({
    "Mouse": [5, 2],
    "Predicted": np.where(y_val_pred == 1, "Field Mouse", "House Mouse"),
    "True":      ["House Mouse", "Field Mouse"]
})
```

```python
#  Aditional: plot the hyperplane
x_values = np.linspace(X_train[:, 0].min() - 5, X_train[:, 0].max() + 5)
y_values = -(w[0] / w[1]) * x_values - (w[2] / w[1])

fig, axes = plt.subplots(1, 2, figsize=(10, 5))

create_plot("Training Data", X_train_plot, y_train, axes[0])
create_plot("Validation Data", X_val_plot, y_val, axes[1])

for i in range(2):
    axes[i].plot(x_values, y_values, color="orange", label="Decision boundary")

plt.tight_layout()
plt.show()
```

3. Is the vectorised Perceptron appropriate for this problem? (0.25 points)

*Yes. The vectorised Perceptron is appropriate for this problem because the dataset is linearly separable — that is, a hyperplane exists that correctly partitions the two classes in the feature space. Since the Perceptron convergence theorem guarantees convergence in a finite number of steps when this condition holds, the algorithm is well-suited for this classification task.*

4. Compute the margin $\gamma$ of your classifier. Check that the number of mistakes satisfies the convergence theorem bound. (0.5 points)

```python
functional_margin = y_train * np.dot(X_train, w)
weight_norm = np.linalg.norm(w[:-1])
geometric_margin = functional_margin / weight_norm

print("Functional margin of the classifier:", functional_margin)
print("Geometrical margin of the classifier:", geometric_margin)

margin = np.min(geometric_margin)
norms = np.linalg.norm(X_train, axis=1)
R = np.max(norms)

bound = (R**2) / (margin**2)

print("Mistakes:", steps, "Theoretical bound: %.2f" % bound, "\nDoes it satisfy the convergence theorem bound?", steps <= bound)
```

5. Normalise the data and re-run. Does convergence change? Why? (0.5 points)

```python
#  no norm
print("Running perceptron without normalising...")
w_z = perceptron(X_train, y_train, max_iter= 1000)
print("=" * 60)

#  min_max norm
X_train_norm = X_train.copy()
X_min = X_train[:, :-1].min(axis=0)
X_max = X_train[:, :-1].max(axis=0)
denom = X_max - X_min
denom[denom == 0] = 1 #  avoid div by 0
X_train_norm[:, :-1] = (X_train[:, :-1] - X_min) / denom

print("Running perceptron with min-max normalization...")
w_m = perceptron(X_train_norm, y_train, max_iter= 1000)
print("=" * 60)

#  z_score
X_train_std = X_train.copy()
mean = X_train[:, :-1].mean(axis=0)
std  = X_train[:, :-1].std(axis=0)
std[std == 0] = 1 #  avoid div by 0
X_train_std[:, :-1] = (X_train[:, :-1] - mean) / std

print("Running perceptron with z-score standarization...")
w_z = perceptron(X_train_std, y_train, max_iter= 1000)
```

*Convergence is significantly faster (from 81 iterations down to 2–3). This is expected given the different scales of the features — ear size is measured in mm and tail length in cm. Additionally, fewer steps are required because the initial weight vector **w** is [0, 0, 0], while the original data ranges from 11 to 25 for ear sizes and 4.5 to 10 for tail lengths. This is precisely why z-score standardization yields better results: although both normalization methods bring all values close to 0 — min-max to the [0, 1] range and z-score to approximately [−1, 1] — z-score centers the values around 0, meaning the model only needs to adjust the orientation of the hyperplane rather than its position, since **w** is initialized to [0, 0, 0] as noted above.*

1. Fit a suitable model to predict `Gusts_Tomorrow`. Report training and validation errors.

```python
X_train = train_data.drop('Gusts_Tomorrow', axis=1)
X_val = valid_data.drop('Gusts_Tomorrow', axis=1)

y_train = train_data['Gusts_Tomorrow']
y_val = valid_data['Gusts_Tomorrow']

model = LinearRegression()

model.fit(X_train, y_train) #  we train a simple linear regression to set a baseline

train_preds = model.predict(X_train)
val_preds = model.predict(X_val)

print("Training MSE:", mean_squared_error(y_train, train_preds))
print("Validation MSE:", mean_squared_error(y_val, val_preds))
```

*As observed, the standard linear model does not generalize well on this dataset: the validation MSE is approximately four times larger than the training MSE, which is indicative of overfitting. To address this, a regularized model will be explored in the following section.*

```python
from sklearn.pipeline import Pipeline
from sklearn.model_selection import GridSearchCV
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.linear_model import Ridge, Lasso

pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('model', Ridge())
])

#  we set a grid of parameters to perform grid search for the best configuration
param_grid = [
    {
        'scaler': [StandardScaler(), MinMaxScaler(), 'passthrough'],
        'model': [Ridge(max_iter=10000), Lasso(max_iter=10000)],
        'model__alpha': [0.001, 0.01, 0.1, 0.5, 1.0, 5.0, 10.0, 15.0, 20.0, 30.0, 40.0, 50.0, 100.0]
    }
]

grid_search = GridSearchCV(
    pipeline, 
    param_grid, 
    cv=5, 
    scoring='neg_mean_squared_error' #  as seen in theory
)

grid_search.fit(X_train, y_train)

best_model = grid_search.best_estimator_
train_preds = best_model.predict(X_train)
val_preds = best_model.predict(X_val)

print("Best Hyperparameters:", grid_search.best_params_)
print("Training MSE:", mean_squared_error(y_train, train_preds))
print("Validation MSE:", mean_squared_error(y_val, val_preds))
```

*Following a grid search over regularization methods, scaling strategies, and penalty terms (alpha), the best-performing model was found to be Ridge regression with α = 30.0 and no feature scaling. Under this configuration, the gap between training and validation MSE is reduced to approximately 2 units — at the cost of a negligible increase of 0.2 units in training error.*

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

pipeline = Pipeline([
    ('poly', 'passthrough'),
    ('scaler', StandardScaler()),
    ('model', LogisticRegression(max_iter=10000))
])

param_grid = [
    {
        'poly': [PolynomialFeatures(degree=2, include_bias=False),
                 PolynomialFeatures(degree=3, include_bias=False)],
        'scaler': [StandardScaler(), MinMaxScaler()],
        'model__C': [0.001, 0.01, 0.1, 0.5, 1.0, 5.0, 10.0, 50.0, 100.0],
        'model__solver': ['lbfgs'],
        'model__class_weight': [None, 'balanced'],
    },
    {
        'poly': ['passthrough'],
        'scaler': [StandardScaler(), MinMaxScaler(), 'passthrough'],
        'model__C': [0.001, 0.01, 0.1, 0.5, 1.0, 5.0, 10.0, 50.0, 100.0],
        'model__solver': ['lbfgs'],
        'model__class_weight': [None, 'balanced'],
    },
]

grid_search = GridSearchCV(
    pipeline,
    param_grid,
    cv=cv,
    scoring='f1_macro'
)

grid_search.fit(X_train, y_train)
best_model = grid_search.best_estimator_

# OOF probs from the best pipeline config (already fitted on full train)
# cross_val_predict refits the pipeline on each fold — correct behaviour
oof_probs = cross_val_predict(
    best_model, X_train, y_train,
    cv=cv, method='predict_proba'
)[:, 1]

# Tune threshold on OOF probs vs true train labels
best_threshold = 0.5
best_f1 = 0

for thresh in np.arange(0.1, 0.9, 0.01):
    preds = (oof_probs >= thresh).astype(int)
    score = f1_score(y_train, preds)
    if score > best_f1:
        best_f1 = score
        best_threshold = thresh

train_probs = best_model.predict_proba(X_train)[:, 1]
val_probs   = best_model.predict_proba(X_val)[:, 1]

train_preds = (train_probs >= best_threshold).astype(int)
val_preds   = (val_probs   >= best_threshold).astype(int)

print("Best Hyperparameters:", grid_search.best_params_)
print(f"Optimal classification threshold (OOF): {best_threshold:.2f}")
print()
print("Training accuracy:", accuracy_score(y_train, train_preds))
print("Validation accuracy:", accuracy_score(y_val, val_preds))
print("Training F1:", f1_score(y_train, train_preds))
print("Validation F1:", f1_score(y_val, val_preds))
print("Validation ROC-AUC:", roc_auc_score(y_val, val_probs))

print("\nValidation classification report:\n")
print(classification_report(y_val, val_preds))

cm = confusion_matrix(y_val, val_preds)

plt.figure(figsize=(5, 4))
sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='YlOrBr',
    xticklabels=['No disorientation', 'Disorientation'],
    yticklabels=['No disorientation', 'Disorientation']
)
plt.xlabel('Predicted label')
plt.ylabel('True label')
plt.title('Confusion Matrix - Validation Set')
plt.tight_layout()
plt.show()
```

*The best configuration found via grid search is a Logistic Regression with C = 0.1 (strong regularization), no polynomial features, no feature scaling, and an optimal classification threshold of 0.31 — lower than the default 0.5, which reflects the model's tendency to assign lower probabilities to the positive class (Disorientation).
Overall, the model achieves a validation accuracy of 69% and a ROC-AUC of 0.72, suggesting moderate discriminative ability. Notably, training and validation metrics are closely aligned, indicating that the model generalizes well and is not overfitting. Looking at per-class performance, the model handles the majority class (No disorientation) more reliably, with both precision and recall of 0.75, whereas the minority class (Disorientation) yields lower precision and recall of 0.60. This is consistent with the confusion matrix, which shows that 6 out of 24 "No disorientation" cases and 6 out of 15 "Disorientation" cases are misclassified — the model thus struggles equally in both directions, though the impact is more significant for the minority class given its smaller support.
In summary, while the model provides a reasonable baseline, its limited capacity (strong regularization, no feature engineering) and the class imbalance present in the dataset suggest there is room for improvement, particularly in the detection of disorientation cases.*

This is reflected in the confusion matrix, where 9 out of 15 disorientation cases are misclassified as no disorientation, representing a high false negative rate that may be particularly undesirable in a clinical context.
Comparing both models, Logistic Regression outperforms the SVM across all key metrics: higher validation accuracy (69% vs. 67%), higher ROC-AUC (0.72 vs. 0.69), and notably superior F1-score for the minority class (0.60 vs. 0.48). Furthermore, the SVM — despite employing balanced class weights — achieves worse recall on the Disorientation class than the Logistic Regression, which relied solely on threshold adjustment. Overall, neither model delivers strong predictive performance, which is likely attributable to the limited dataset size and the moderate separability of the classes; nevertheless, Logistic Regression stands as the more suitable choice between the two.*
